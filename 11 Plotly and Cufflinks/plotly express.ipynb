```python
df3 = pd.DataFrame({'x':[1,2,3], 'y':[10,20,30],'z':[100,200,300]})
```

```python
df3
import plotly.graph_objects as go
```

```python
go.Figure(data=[go.Surface(z=df3['z'], x=df3['x'], y=df3['y'])])
```

```python


x = [1, 2, 3]
y = [10, 20, 30]

# Create a 2D grid of z-values (e.g. z = x * y)
X, Y = np.meshgrid(x, y)
Z = X * Y  # or any 2D function you want

fig = go.Figure(data=[go.Surface(x=X, y=Y, z=Z)])
fig.show()
```
