```python
x=['A','B','C'] 
y=[4,7,3]
men = go.Bar(x=x,y=y,name='Men')
x=['A','B','C']
y=[5,6,4]
women = go.Bar(x=x,y=y,name='Women')

fig = go.Figure(data=[men,women])
fig.show()
```

```python
fig.update_layout(title=dict(text = 'Compariosn by gender'),xaxis_title='Category',yaxis_title='Values')
fig.show()
```

```python
x = np.linspace(0,2*np.pi,10)
```

```python
scatter = go.Scatter(x= np.sin(x),y=np.cos(x))
fig = go.Figure(data=[scatter])
fig.show()
```
