```python
# ==============================================================
# 📗 Section 2: Pandas Practice (5 Questions)
# ==============================================================

import pandas as pd
import seaborn as sns

# Load a sample dataset for practice
tips = sns.load_dataset('tips')

# Q6. Display the first 5 rows of the tips dataset.
tips.head()


# Q7. Show how many rows and columns are in the dataset.
tips.info()


# Q8. Calculate the average total_bill per day.
print(tips.groupby('day')['total_bill'].mean())

# Q9. Find the total number of missing values in the dataset.
print(tips.isnull().sum() )# -> sum of all true values


# Q10. Create a new column called 'tip_percentage' = (tip / total_bill) * 100
tips['tip_percentage'] = (tips['tip'] / tips['total_bill']) * 100
tips.head()
```

```python
import matplotlib.pyplot as plt

# Q11. Create a simple line plot for x = [1,2,3,4,5] and y = [2,4,6,8,10].
x = [1,2,3,4,5]  
y = [2,4,6,8,10]
plt.plot(x,y)
plt.show()

# Q12. Add a title and axis labels to your previous plot.
x = [1,2,3,4,5]  
y = [2,4,6,8,10]
plt.plot(x,y)
plt.title('My Picture')
plt.xlabel(' X axis')
plt.ylabel(' Y axis ')
plt.show()


# Q13. Create a bar chart showing total_bill by day using the tips dataset.
total_bill = tips.groupby('day')['total_bill'].sum()
# print(total_bill)
plt.bar(total_bill.index, total_bill.values)
plt.show()

# Q14. Create a histogram of the 'total_bill' column with 20 bins.
plt.hist(tips['total_bill'],bins=20)
plt.show()

# Q15. Create two subplots (1 row, 2 columns) showing 'tip' and 'total_bill' histograms.
plt.subplot(1,2,1)
plt.hist(tips['tip'])
plt.subplot(1,2,2)
plt.hist(tips['total_bill'])
plt.show()
```

```python
# ==============================================================
# 📘 Section 4: Seaborn Practice (5 Questions)
# ==============================================================
import seaborn as sns
# Q16. Load the 'titanic' dataset and show its first 5 rows.
titanic = sns.load_dataset('titanic')
print(titanic.head())
# Q17. Create a countplot showing the number of male and female passengers.
sns.countplot(x= titanic['sex'] , data= titanic )
plt.show()

# Q18. Create a boxplot showing the distribution of 'age' by 'class'.
sns.boxplot(x='class',y = 'age',data = titanic)
plt.show()


# Q19. Create a heatmap showing the correlation between numeric columns of the Titanic dataset.
corr_value = titanic.corr()
sns.heatmap(data=corr_value)
plt.show()

# Q20. Create a scatter plot showing relationship between 'total_bill' and 'tip' from the tips dataset.
sns.scatterplot(x='total_bill', y= 'tip',data = tips)
plt.show()
```
