1️⃣ NumPy Basics

Concept:

Efficient arrays (1D, 2D, N-Dimensional) and fast numerical operations.

Arrays are faster than Python lists.

Operations: arithmetic, indexing, slicing, reshaping.

Aggregate functions: mean, sum, std, min, max.

Code Example:

```python
import numpy as np


# Create 1D array
arr = np.array([1,2,3,4,5])
print(arr)


# Array shape and dtype
print(arr.shape, arr.dtype)


# Arithmetic operations
print(arr * 2)
print(arr + 10)


# Slicing
print(arr[1:4])


# Aggregate functions
print('Mean:', arr.mean())
print('Sum:', arr.sum())
print('Std:', arr.std())


# Reshaping
arr2d = np.arange(1,10).reshape(3,3)
print(arr2d)
```

# 2️⃣ Pandas Essentials

Concept:

DataFrames (2D table) and Series (1D).

Functions: head(), info(), describe(), filtering, grouping, missing values.

```python
import pandas as pd
import seaborn as sns


# Load dataset
tips = sns.load_dataset('tips')


# Inspect data
print(tips.head())
print(tips.info())
print(tips.describe())


# Filter data
print(tips[tips['tip'] > 5])


# Grouping & aggregation
print(tips.groupby('day')['total_bill'].mean())


# Missing values
print(tips.isnull().sum())


# Add new column
tips['tip_percentage'] = tips['tip'] / tips['total_bill'] * 100
print(tips.head())
```

# 3️⃣ Matplotlib Visualization

Concept:

Static plots: line, bar, histogram, scatter, subplots.

Customize with title, labels, legends, color, linewidth.

Code Example:

```python
import matplotlib.pyplot as plt


# Line plot
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()


# Bar chart
total_bill_per_day = tips.groupby('day')['total_bill'].sum()
plt.bar(total_bill_per_day.index, total_bill_per_day.values, color='green')
plt.show()


# Histogram
plt.hist(tips['total_bill'], bins=20, color='blue')
plt.show()


# Subplots
plt.subplot(1,2,1)
plt.hist(tips['tip'], bins=15)
plt.subplot(1,2,2)
plt.hist(tips['total_bill'], bins=15)
plt.show()
```

# 4️⃣ Seaborn Visualization

Concept:

Builds statistical plots on top of Matplotlib.

Plots: countplot, boxplot, heatmap, scatterplot.

Supports hue, style, col for grouping.

Automatic aggregation for bar plots.

```python
# Load dataset
titanic = sns.load_dataset('titanic')


# Countplot
sns.countplot(x='sex', data=titanic)
plt.show()


# Boxplot
sns.boxplot(x='class', y='age', data=titanic)
plt.show()


# Heatmap of correlations
corr = titanic.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()


# Scatterplot
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.show()


# Barplot with sum aggregation
sns.barplot(x='day', y='total_bill', data=tips, estimator=sum)
plt.show()
```
