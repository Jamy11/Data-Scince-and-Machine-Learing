## 🟨 NumPy Intermediate Practice (Questions 6–10)

6.  Given `arr = np.array([10, 20, 30, 40, 50])`, extract the **first 3 elements** using slicing.

7.  Create a 3×3 array and find:
- The **mean**
- The **maximum**
- The **minimum**

8. Given `arr = np.array([[1, 2], [3, 4], [5, 6]])`, reshape it into a **2×3** array.

9. Generate an array of 10 random numbers and find the **indices of the minimum and maximum values**.

10. Create an array of 10 random integers and replace all values **greater than 50 with 50**.


```python
arr = np.array([10, 20, 30, 40, 50])
print( arr[:3] )

random_array = np.random.randint(0,100,(3,3))
print(f'random array = {random_array}')
print( 'Next Answer is 7' )
print( np.mean(random_array) )
print( np.max(random_array) )
print( np.min(random_array) )

print( 'Next Answer is 8' )
arr = np.array([[1, 2], [3, 4], [5, 6]])
print( arr.reshape(2,3) )

print( 'Next Answer is 9' )
random_array = np.random.randint(0,100,10)
print(f'random array = {random_array}')
print( random_array.argmax() )
print( np.argmin(random_array) )

print( 'Next Answer is 10' )

print( random_array[ random_array>50 ] )
random_array[ random_array>50 ] = 50
print(random_array)
```
