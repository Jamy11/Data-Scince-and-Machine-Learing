1. **Simple Line Plot**  
   Create a line plot showing `x = [0,1,2,3,4]` and `y = [0,1,4,9,16]`.

2. **Add Labels and Title**  
   Add an **x-axis label**, **y-axis label**, and **title** to your previous plot.

3. **Line Style and Color**  
   Modify your line plot to make the line **dashed** and **red** in color.

4. **Multiple Lines in One Plot**  
   Plot two lines in the same figure:  
   - `x = [0,1,2,3,4]`, `y1 = [0,1,4,9,16]`  
   - `y2 = [0,1,2,3,4]`  
   Add a **legend**.




```python
x = [0,1,2,3,4]
y = [0,1,4,9,16]

plt.plot(x,y,color='red',ls='--')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Yo title')
plt.show()

x = [0,1,2,3,4]
y1 = [0,1,4,9,16]
y2 = [0,1,2,3,4]

plt.plot(x,y1, color='red',label='Line 1')
plt.plot(x,y2, color='blue',label='Line 2')
# plt.plot(y1,y2, color='orange',label='Line 3')
# plt.plot(y2,y1, color='black',label='Line 4')
plt.legend()
plt.show()
```

## 🟩 Section 2: Subplots and Customization

5. **Subplots**  
   Create two subplots (1 row, 2 columns):  
   - Left: sine curve  
   - Right: cosine curve  
   Use `numpy.linspace()` for x-values between 0 and 2π.

6. **Figure Size and Style**  
   Create a plot with custom **figure size (8,4)** and **grid lines**.

7. **Bar Plot**  
   Plot a bar chart showing:

`Categories = ['A', 'B', 'C', 'D']`

`Values = [10, 20, 15, 30]`

```python
x = np.linspace(0,2*np.pi,10)

fig , axis = plt.subplots(1,2,figsize=(8,4))

axis[0].plot(x,np.sin(x))
axis[1].plot(x,np.cos(x))

plt.grid(True)
plt.tight_layout()
plt.show()

Categories = ['A', 'B', 'C', 'D']

Values = [10, 20, 15, 30]
plt.bar(Categories,Values)
plt.show()
```

## 🟥 Section 3: Advanced Visualization


8. **Histogram**

9. **Scatter Plot with Style**  
Create a scatter plot for:

Customize with **green markers**, size `s=100`, and alpha transparency `0.5`.

`x = [1,2,3,4,5]`

`y = [5,4,3,2,1]`


```python
x= np.random.randint(0,100,100)

plt.hist(x,bins=15)
plt.show()

x = [1,2,3,4,5]
y = [5,4,3,2,1]
plt.scatter(x,y, color='green', marker='*',s=100, alpha=0.5)
plt.show()
```
