```python
df = sns.load_dataset('titanic')
df.head()
```

```python
sns.countplot(data=df,x='sex')
```

```python
sns.barplot(data=df,x='class',y='fare')
```

```python
sns.histplot(df['age'],bins=20)
```

```python
sns.boxplot(x='class',y='age',data=df)
```

```python
sns.scatterplot(x='class',y='age',data=df,hue='sex')
plt.show()
sns.stripplot(x='class',y='age',data=df,hue='sex')
plt.show()
sns.swarmplot(x='class',y='age',data=df,hue='sex')
plt.show()
```

```python
df2 = sns.load_dataset('iris')
sns.pairplot(data=df2)
```

```python
g = sns.FacetGrid(df, col='class', row='sex')
g.map(sns.histplot,'age')
```

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]
sns.lineplot(x=x,y=y,color='red',marker='*',size=2)
```

```python
print(df['age'].isnull().sum())
print(df['embarked'].isnull().sum())
```

```python
df['age'].fillna(df['age'].mean())
df['embarked'].fillna('S')
```
