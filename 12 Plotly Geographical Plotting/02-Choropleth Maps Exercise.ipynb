## Plotly Imports

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

** Import pandas and read the csv file: 2014_World_Power_Consumption**

```python
df = pd.read_csv('2014_World_Power_Consumption')
```

```python
df.head()
```

** Check the head of the DataFrame. **

** Referencing the lecture notes, create a Choropleth Plot of the Power Consumption for Countries using the data and layout dictionary. **

```python

data = go.Choropleth(
    locations=df['Country'],
    text=df['Text'],
    z=df['Power Consumption KWH'],
    locationmode='country names',
)
layout= go.Layout(
    title='Power consumptions',
    geo = dict(
        showframe= False,
        showcoastlines=True,
        projection_type = 'natural earth'
    )
)
```

```python
choromap = go.Figure(data = [data],layout = layout)
choromap.show(validate=False)
```

** Import the 2012_Election_Data csv file using pandas. **

```python
df2 = pd.read_csv('2012_Election_Data')
```

** Check the head of the DataFrame. **

** Now create a plot that displays the Voting-Age Population (VAP) per state. If you later want to play around with other columns, make sure you consider their data type. VAP has already been transformed to a float for you. **

```python
# df2['Voting-Eligible Population (VEP)'] = (
#     df2['Voting-Eligible Population (VEP)'].str.replace(',', '').astype(float)             # convert to numeric
# )

# df2['State Abv'] = df2['State Abv'].replace('District of Columbia', 'DC')
data = go.Choropleth(
    locations = df2['State Abv'],
    locationmode = 'USA-states',
    z = df2['Voting-Age Population (VAP)'],
    text = df2['State'],
    colorscale = 'YlOrRd',        # add a color scale
    colorbar_title = 'Eligible Voters'
)
layout = go.Layout(
    title='Voting-Age Population',
    geo= dict(
        scope='usa',
    )
)
```
