```python
import pandas as pd
import numpy as np
%matplotlib inline
import plotly.express as px
import plotly.graph_objects as go
```

```python
data = dict(type = 'choropleth',
            locations = ['AZ','CA','NY'],
            locationmode = 'USA-states',
            colorscale= 'Portland',
            text= ['text1','text2','text3'],
            z=[1.0,2.0,3.0],
            colorbar = {'title':'Colorbar Title'})
```

```python
layout = dict(geo={'scope':'usa'})
print(layout)
```

```python
# Create the figure
fig = go.Figure(data=[data], layout=layout)

# Show the figure
fig.show()
```

```python
df = pd.read_csv('2011_US_AGRI_Exports')
```

```python

df.head()
```

```python
data = dict(type = 'choropleth',
            colorscale = 'YIOrRd',
            locations =df['code'],
            locationmode = 'USA-states',
            z = df['total exports'],
            text= df['text'],
            colorbar = {'title':'Millions USD'},
            marker = dict(line = dict(color = 'rgb(255,255,255)', width = 2))
           
           )
```

```python
choropleth = go.Choropleth(
    colorscale='YlOrRd',
    locations=df['code'],
    locationmode='USA-states',
    z=df['total exports'],
    text=df['text'],
    colorbar_title='Millions USD',
    marker=dict(line=dict(color='rgb(255,255,255)', width=2))
)

# Layout
layout = go.Layout(
    title='2011 US Agriculture Exports',
    geo=dict(
        scope='usa',
        showlakes=True,
        lakecolor='rgb(85,173,240)'
    )
)
```

```python
Fig = go.Figure(data=[choropleth],layout=layout)
Fig.show()
```

# Plotting the whole world

```python
df2 = pd.read_csv('2014_World_GDP')
```

```python
df2.head()
```

```python
data = go.Choropleth(    
    # colorscale='YlOrRd',
    locations=df2['CODE'],
    z=df2['GDP (BILLIONS)'],
    # locationmode='ISO-3',
    text=df2['COUNTRY'],
    colorbar_title='Billions USD',
    marker=dict(line=dict(color='rgb(255,255,255)', width=.5)))

layout = go.Layout(
    title='GDP Spending (Global)',
    geo=dict(
        showframe=False,
        projection_type='natural earth'   # Best for global view
    )
)
```

```python
Fig2 = go.Figure(data=[data],layout=layout)
Fig2.show()
```
